## Approximation

$\textbf{Marginalization:}$

$\textcolor{blue}{\mathbb{p}_{\theta}(x)}= \int_{z}{\mathbb{p}_{\theta}(x,y)}$ 

$\textbf{Conditional Probability:}$

$\textcolor{darkred}{\mathbb{p}_{\theta}(x)}= \frac{\mathbb{p}_{\theta}(x,y)}{\textcolor{darkred}{\mathbb{p}_{\theta}(y | x)}}$ 

Both $\textcolor{darkred}{\mathbb{p}_{\theta}(y | x)}$ and $\textcolor{darkred}{\mathbb{p}_{\theta}(x)}$ are intractable. But if we reached one we find the other...

So let's approximate using $D_{kl}(\textcolor{green}{q_{\phi}(y | x)} || \textcolor{darkred}{p_{\theta}(y | x)})=\int_{z} \textcolor{green}{q_{\phi}(y | x)} log(\frac{q_{\phi}(y | x)}{\textcolor{darkred}{p_{\theta}(y | x)}}) d_{z}$
